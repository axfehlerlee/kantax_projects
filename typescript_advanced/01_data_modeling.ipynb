# 문항 1 · 회원과 상품 모델링

회원과 상품은 숫자 ID를 공유합니다. 역할과 카테고리는 문자열 리터럴 유니언으로 제한합니다.

```typescript
interface Identifiable {
    id: number;
}

type UserRole = 'admin' | 'customer' | 'guest';
type Category = 'electronics' | 'clothing' | 'books';

interface User extends Identifiable {
    name: string;
    email: string;
    role: UserRole;
    isActive: boolean;
}

interface Product extends Identifiable {
    title: string;
    price: number;
    category: Category;
    stock: number;
    discountRate?: number;
}

const users: User[] = [
    { id: 1, name: '지민', email: 'jimin@example.com', role: 'admin', isActive: true },
    { id: 2, name: '수현', email: 'suhyeon@example.com', role: 'customer', isActive: true },
    { id: 3, name: '민준', email: 'minjun@example.com', role: 'guest', isActive: false },
    { id: 4, name: '하윤', email: 'hayun@example.com', role: 'customer', isActive: false },
];

const products: Product[] = [
    { id: 101, title: '키보드', price: 50000, category: 'electronics', stock: 5, discountRate: 0.2 },
    { id: 102, title: '셔츠', price: 30000, category: 'clothing', stock: 3 },
    { id: 103, title: 'TypeScript 입문', price: 25000, category: 'books', stock: 8, discountRate: 0 },
];
console.log(`모델 확인 · 회원 ${users.length}명, 상품 ${products.length}종`);
```

## 역할 필터

역할과 활성 여부는 별개입니다. 같은 역할의 회원은 비활성 상태여도 결과에 포함합니다.

```typescript
function filterUsersByRole(users: User[], targetRole: UserRole): User[] {
    return users.filter(user => user.role === targetRole);
}

const customers: User[] = filterUsersByRole(users, 'customer');
if (customers.map(user => user.id).join(',') !== '2,4') {
    throw new Error('고객 필터 결과가 예상과 다릅니다.');
}
console.log('고객:', customers.map(user => user.name).join(', '));
console.log('빈 목록:', filterUsersByRole([], 'admin').length);
```

## 총 결제 금액

할인율은 소수로 표현합니다. 예를 들어 0.2는 20% 할인입니다. 할인율 생략 시 정가를 적용하고, 배열의 각 상품을 한 개씩 합산합니다.

```typescript
function calculateTotalPrice(products: Product[]): number {
    return products.reduce((total, product) => {
        const discountRate: number = product.discountRate ?? 0;
        return total + product.price * (1 - discountRate);
    }, 0);
}

const totalPrice: number = calculateTotalPrice(products);
if (totalPrice !== 95000 || calculateTotalPrice([]) !== 0) {
    throw new Error('총 결제 금액이 예상과 다릅니다.');
}
console.log(`총 결제 금액: ${totalPrice.toLocaleString('ko-KR')}원`);
console.log('빈 상품 목록: 0원');
```
